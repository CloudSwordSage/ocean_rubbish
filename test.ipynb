```python
img = cv2.imread('./data/Bistrina/Bluerobotics HD/1.jpg')

x, y, w, h = data['annotations'][0]['bbox']
cw = x + w / 2
ch = y + h / 2
img = cv2.rectangle(img, (int(cw-w/2), int(ch-h/2)), (int(cw+w/2), int(ch+h/2)), (0, 255, 0), 2)
x, y, w, h = data['annotations'][1]['bbox']
cw = x + w / 2
ch = y + h / 2
img = cv2.rectangle(img, (int(cw-w/2), int(ch-h/2)), (int(cw+w/2), int(ch+h/2)), (0, 255, 0), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
```
